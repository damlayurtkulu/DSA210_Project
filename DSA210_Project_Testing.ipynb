```python
import pandas as pd
from scipy.stats import pearsonr, chi2_contingency
```

```python
df = pd.read_csv("merged_clean_data.csv")
df.head()
```

```python
# H1: PEARSON CORRELATION

corr_h1, p_h1 = pearsonr(
    df["avg_unemployment_rate"],
    df["purchase_amount_(usd)"]
)

print("H1 Correlation:", corr_h1)
print("H1 P-value:", p_h1)
```

```python
freq_map = {
    "Daily": 30,
    "Weekly": 4,
    "Bi-Weekly": 2,
    "Fortnightly": 2,
    "Monthly": 1,
    "Quarterly": 0.33,
    "Semi-Annually": 0.16,
    "Annually": 0.083
}

df["purchase_frequency_numeric"] = df["frequency_of_purchases"].map(freq_map)
```

```python
# H2: PEARSON CORRELATION


df_h2 = df.dropna(subset=["purchase_frequency_numeric"])

corr_h2, p_h2 = pearsonr(
    df_h2["avg_unemployment_rate"],
    df_h2["purchase_frequency_numeric"]
)

print("H2 Correlation:", corr_h2)
print("H2 P-value:", p_h2)
```

```python
#H3: CHI-SQUARE TEST


discount_table = pd.crosstab(
    df["unemployment_group"],
    df["discount_applied"]
)

chi2_h3, p_h3, dof_h3, exp_h3 = chi2_contingency(discount_table)

print("H3 Chi-square:", chi2_h3)
print("H3 P-value:", p_h3)
```

```python
# H4: CHI-SQUARE TEST


subscription_table = pd.crosstab(
    df["unemployment_group"],
    df["subscription_status"]
)

chi2_h4, p_h4, dof_h4, exp_h4 = chi2_contingency(subscription_table)

print("H4 Chi-square:", chi2_h4)
print("H4 P-value:", p_h4)
```
